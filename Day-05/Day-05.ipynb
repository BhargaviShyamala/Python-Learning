# Conditional Statements

```python
#Conditional statements are used to make decisions in a Python program.
#They check whether a condition is True or False and execute code based on the result.
#Types of Conditional Statements:if statement, Short-hand if, if-else statement, if-elif-else statement, Nested if-else, Conditional Expression (Ternary Operator),match-case statement
```

# if Statement

```python
# The if statement executes code only when the condition is True.
age = 20
if age >= 18:
    print("Eligible to vote.")
```

# Short-Hand if

```python
# When there is only one statement, we can write it in one line.
age = 19
if age >= 18: print("Eligible to vote.")
```

# if-else Statement

```python
# if-else is used when there are two possible choices.
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

# if-elif-else

```python
# Use elif when you have multiple conditions.
age = 25
if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
```

# Nested if-else

```python
# A nested if means putting one if statement inside another if statement.
age = 70
is_member = True
if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
```

# Conditional Expression / Ternary Operator

```python
# It is a short way of writing if-else in one line.
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)
```

# match-case

```python
# match-case is similar to switch-case in some other programming languages.
number = 2
match number:
    case 1:
        print("One")
    case 2 | 3:
        print("Two or Three")
    case _:
        print("Other number")
```

# Loops

```python
# A loop is used to execute the same block of code repeatedly.
# For example, if we want to print "Hello" 5 times, instead of writing print() 5 times, we can use a loop.
#Main Types of Loops
       # For loop – used when we want to iterate over a sequence or a known range.
       # While loop – used when we want to repeat code as long as a condition is True.
```

# For Loop

```python
# A for loop is used to iterate through items in a list, tuple, string, or range.
n = 4
for i in range(0, n):
    print(i)
```

# For Loop with a List

```python
#  We can directly iterate through the elements of a list.
fruits = ["Apple", "Banana", "Mango"]
for fruit in fruits:
    print(fruit)
```

# For Loop Using Index

```python
# We can access list elements using their index.
a = ["geeks", "for", "geeks"]
for idx in range(len(a)):
    print(a[idx])
```

# While Loop

```python
# A while loop repeatedly executes code as long as the condition is True.
cnt = 0
while cnt < 3:
    cnt = cnt + 1
    print("Hello Geek")
```

# Infinite While Loop

```python
# An infinite loop is a loop that never stops because its condition always remains True.
while True:
    print("Hello")
```
