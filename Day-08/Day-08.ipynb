# Python Tuples

```python
# A tuple is an ordered and immutable collection of elements.
# Ordered → Elements have a fixed position.
# Immutable → Once created, we cannot change, add, or remove individual elements.
# Heterogeneous → A tuple can contain different data types.
# Tuples are written using parentheses ().
```

### Creating a Tuple


```python
# Empty Tuple
tup = ()
print(tup)
```

```python
# Tuple with Strings
tup = ("Bhargavi", "Shyamala")
print(tup)
```

```python
# Using tuple() Function
tup = tuple("Bhargavi")

print(tup)
```

### Tuple with Different Data Types

```python
# A tuple can contain different types of values.
tup = (5, "Welcome", 7.5, True, [1, 2, 3])
print(tup)
```

```python
# Accessing Tuple ElementsAccessing Tuple Elements:
    # We can access tuple elements using indexing.
    # Index starts from 0.
tup = ("Python", "Java", "C++", "SQL")
print(tup[0])
print(tup[1])
print(tup[2])
```

### Negative Indexing

```python
# Negative indexing starts from the last element.
tup = ("Python", "Java", "C++", "SQL")
print(tup[-1])
print(tup[-2])
```

### Tuple Slicing

```python
# Slicing means taking a portion of a tuple.
# Syntax: tuple[start:stop]
tup = ("Python", "Java", "C++", "SQL")
print(tup[1:3])
```

### Tuple Unpacking

```python
# Tuple unpacking means assigning tuple values to separate variables.
tup = ("Python", "Java", "SQL")
a, b, c = tup
print(a)
print(b)
print(c)
```

### Concatenation of Tuples

```python
# We can combine tuples using the + operator.
tup1 = (1, 2, 3)
tup2 = (4, 5, 6)
tup3 = tup1 + tup2
print(tup3)
```

### Tuple Slicing with Step

```python
# Syntax: tuple[start:stop:step]
tup = tuple("PYTHON")
print(tup[1:])
print(tup[::-1])
print(tup[::2])
```

### Deleting a Tuple

```python
# Because tuples are immutable, we cannot delete individual elements
tup = (1, 2, 3)
del tup
print(tup)
```

### Tuple Unpacking with *


```python
# The * operator collects multiple values into a list.
tup = (1, 2, 3, 4, 5)
a, *b, c = tup
print(a)
print(b)
print(c)
```

# Python Dictionary

```python
# A dictionary stores information in key : value format.
data = {
    "name": "Jake",
    "age": 22
}
print(data)
```

###  Creating a Dictionary

```python
# We use { } to create a dictionary:
data = {
    "name": "Jake",
    "age": 22
}
print(data)
```

### Dictionary Keys and Values

```python
student = {
    "name": "Bhargavi",
    "age": 22,
    "course": "Python"
}
print(student)
```

### Accessing Dictionary Items

```python
# We can access a value using its key.
student = {
    "name": "Bhargavi",
    "age": 22
}
print(student["name"])
print(student["age"])
```

```python
# Using get(): We can also use the get() method.
student = {
    "name": "Bhargavi",
    "age": 22
}

print(student.get("name"))
print(student.get("age"))
```

```python
# [] vs get():
student = {
    "name": "Bhargavi"
}
print(student["age"])
```

```python
student = {
    "name": "Bhargavi"
}

print(student.get("age"))
```

```python
# So: dictionary[key] → gives KeyError if key doesn't exist.
#    dictionary.get(key) → returns None if key doesn't exis
```

### Adding a New Item

```python
# We can add a new key-value pair using =.
student = {
    "name": "Bhargavi"
}
student["age"] = 22
print(student)
```

### Updating a Dictionary Item

```python
# If the key already exists, its value is changed.
student = {
    "name": "Bhargavi",
    "age": 22
}
student["age"] = 23
print(student)
```

### Removing Items from Dictionary

```python
# del: del removes an item using its key.
student = {
    "name": "Bhargavi",
    "age": 22
}

del student["age"]
print(student)
```

```python
# popitem(): popitem() removes and returns the last inserted key-value pair.
student = {
    "name": "Bhargavi",
    "age": 22
}
print(student.popitem())
```

```python
# clear(): clear() removes all items.
student = {
    "name": "Bhargavi",
    "age": 22
}
student.clear()
print(student)
```

### Iterating Through a Dictionary

```python
# Iteration means going through dictionary items one by one.
# We commonly use a for loop.
```

```python
# Iterate Through Keys:
student = {
    "name": "Bhargavi",
    "age": 22
}
for key in student:
    print(key)
```

### Iterating Through Values

```python
# Use the values() method.
student = {
    "name": "Bhargavi",
    "age": 22
}
for value in student.values():
    print(value)
```

### Iterating Through Keys and Values

```python
# Use the items() method.
student = {
    "name": "Bhargavi",
    "age": 22
}

for key, value in student.items():
    print(key, value)
```

### Nested Dictionary

```python
# A nested dictionary is a dictionary inside another dictionary.
student = {
    "student": {
        "name": "Bhargavi",
        "age": 22
    }
}
print(student)
```

### Dictionary with Different Data Types

```python
# Dictionary values can contain different data types.
data = {
    "name": "Bhargavi",
    "age": 22,
    "marks": 85.5,
    "passed": True,
    "subjects": ["Python", "SQL"]
}
print(data)
```
