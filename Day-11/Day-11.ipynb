# Python OOP Concepts

```python
# OOP = Object-Oriented Programming
# OOP is a way of writing programs using classes and objects.
# For example, think about a Car:
  # Car has data → color, brand, price
  #Car has actions → start(), stop(), drive()
#In Python, we can represent this using a class.
```

### Class

```python
# A class is a blueprint/template for creating objects.
class Dog:
    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age


# Creating objects
dog1 = Dog("Buddy", 3)
dog2 = Dog("Rocky", 5)

# Printing object details
print("Dog 1 Name:", dog1.name)
print("Dog 1 Age:", dog1.age)
print("Dog 1 Species:", dog1.species)

print()

print("Dog 2 Name:", dog2.name)
print("Dog 2 Age:", dog2.age)
print("Dog 2 Species:", dog2.species)
```

### __init__() Constructor

```python
# __init__() is a special method that runs automatically when an object is created.
class Dog:

    def __init__(self, name, age):
        self.name = name
        self.age = age


dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.age)
```

```python
# What is self?
# self refers to the current object.
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


s1 = Student("Ravi", 20)
s2 = Student("Priya", 21)

print(s1.name)
print(s1.age)

print(s2.name)
print(s2.age)
```

### Object

```python
# An object is an instance of a class.
class Dog:
    pass

dog1 = Dog()

print(dog1)
print(type(dog1))
```

```python
# Creating an Object
class Dog:

    species = "Canine"

    def __init__(self, name, age):
        self.name = name
        self.age = age


dog1 = Dog("Buddy", 3)

print(dog1.name)
print(dog1.age)
print(dog1.species)
```

### Methods

```python
# A method is a function defined inside a class.
class Dog:

    def __init__(self, name):
        self.name = name

    def bark(self):
        print(self.name, "is barking")
dog1 = Dog("Buddy")

dog1.bark()
```
