# Python Keywords

```python
#A keyword is a reserved word in Python that has a special meaning.
#We cannot use keywords as variable names, function names, or class names.
```

```python
# Get All Python Keywords
import keyword
print("Python Keywords:")
print(keyword.kwlist)
```

```python
# Keywords vs Identifiers
   #Keywords:
      Reserved by Python.
      Have predefined meanings.
      Cannot be used as variable names.
      Examples: if, for, while, class.
   # Identifiers:
      Names created by the programmer.
      Used for variables, functions, classes, etc.
      Examples: age, name, student, calculate_sum
```

```python
age = 20
if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not eligible to vote")
```

# Data Types

```python
# Data Type tells Python what kind of data a variable stores.
age = 25
name = "Bhargavi"
marks = 85.5
print(type(age))
print(type(name))
print(type(marks))
```

```python
#Integer (int): Stores whole numbers.
age = 25
print(age)
print(type(age))
```

```python
# Float (float): Stores decimal numbers.
price = 99.50
print(price)
print(type(price))
```

```python
# Complex (complex): Stores a number with real and imaginary parts.
number = 2 + 3j
print(number)
print(type(number))
```

```python
#String (str): Stores text.
name = "Python"
print(name)
print(type(name))
print(name[0])
```

```python
# List (list): A list stores multiple values and can be changed.
marks = [80, 90, 75]
print(marks)
print(marks[1])
marks[1] = 95
print(marks)
```

```python
#Tuple (tuple): A tuple stores multiple values but cannot be changed.
numbers = (10, 20, 30)
print(numbers)
print(numbers[1])
print(type(numbers))
```

```python
# Important difference:
         List   → [ ] → Changeable
         Tuple  → ( ) → Not changeable
```

# Boolean (bool)

```python
# Boolean has only two values: True and False
age = 20
print(age > 18)
print(type(age > 18))
```

```python
# Set (set): A set stores unique values.
numbers = {10, 20, 10, 30, 20}
print(numbers)
print(type(numbers))
```

```python
# Dictionary (dict): A dictionary stores data as key-value pairs.
student = {
    "name": "Alex",
    "age": 22,
    "marks": 85
}

print(student["name"])
print(student["age"])
```
