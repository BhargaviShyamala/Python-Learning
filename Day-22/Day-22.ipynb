# Printing Pyramid Patterns in Python

### Full Pyramid

```python
# Print a full pyramid pattern

def full_pyramid(n):
    # Loop through each row
    for i in range(1, n + 1):

        # Print spaces before stars
        for j in range(n - i):
            print(" ", end="")

        # Print stars
        for k in range(1, 2 * i):
            print("*", end="")

        # Move to the next line
        print()


# Call the function
full_pyramid(5)
```

### Number Pyramid

```python
# Print a number pyramid

def print_number_pyramid(rows):
    # Loop through each row
    for i in range(1, rows + 1):

        # Print spaces
        for j in range(rows - i):
            print(" ", end="")

        # Print numbers
        for j in range(2 * i - 1):
            print(j + 1, end="")

        # Move to the next line
        print()


# Call the function
print_number_pyramid(5)
```

### Half Pyramid

```python
# Print a half pyramid using stars

def half_pyramid(n):
    # Loop through each row
    for i in range(1, n + 1):

        # Print stars
        for j in range(1, i + 1):
            print("* ", end="")

        # Move to the next line
        print()


# Call the function
half_pyramid(5)
```

### Number Half Pyramid

```python
# Print a half pyramid using numbers

def numpat(n):
    # Loop through each row
    for i in range(1, n + 1):

        # Start number from 1
        num = 1

        # Print numbers according to the row
        for j in range(1, i + 1):
            print(num, end=" ")
            num += 1

        # Move to the next line
        print()


# Call the function
numpat(5)
```

### Inverted Half Pyramid

```python
# Print an inverted half pyramid

def inverted_half_pyramid(n):
    # Start from n and decrease to 1
    for i in range(n, 0, -1):

        # Print stars
        for j in range(1, i + 1):
            print("* ", end="")

        # Move to the next line
        print()


# Call the function
inverted_half_pyramid(5)
```

### Hollow Pyramid

```python
# Print a hollow pyramid

def hollow_pyramid(n):
    # Loop through each row
    for i in range(1, n + 1):

        # Loop through positions in the row
        for j in range(1, 2 * n):

            # Print star at the boundary or bottom row
            if j == n - i + 1 or j == n + i - 1 or i == n:
                print("*", end="")
            else:
                # Print spaces inside the pyramid
                print(" ", end="")

        # Move to the next line
        print()


# Call the function
hollow_pyramid(5)
```
