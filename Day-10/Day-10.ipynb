# Python Arrays

```python
# An array is a collection of multiple elements stored together. In Python, there are three common ways to work with array-like data:
   # Python Lists
   # array module
   # NumPy arrays
```

### Python List

```python
# Python does not have a traditional built-in array like C or Java. Lists are commonly used instead.
# A list can store different types of data.
a = [1, "Hello", 3.14]
print(a)
```

```python
a.append(20)
print(a)
```

### Python array Module

```python
# Python provides an array module for creating a typed array.
import array
a = array.array('i', [1, 2, 3, 4])
print(a)
```

### Why Use an Array?

```python
# Unlike a list, a Python array normally stores elements of the same type.
import array
a = array.array('i', [10, 20, 30])
print(a)
```

### Array Typecodes

```python
# The typecode tells Python what type of values the array contains.
import array
a = array.array('i', [10, 20, 30])
print(a)
```

```python
# Float array
import array
a = array.array('f', [1.5, 2.5, 3.5])
print(a)
```

### Accessing Array Elements

```python
# Array elements are accessed using index numbers.
#Index starts from 0.
import array
a = array.array('i', [10, 20, 30, 40])
print(a[0])
print(a[2])
```

### Adding Elements — append()

```python
# append() adds an element to the end.
import array
a = array.array('i', [10, 20, 30])
a.append(40)
print(a)
```

### Adding Elements — insert()

```python
# insert() adds an element at a specific index.
import array
a = array.array('i', [10, 20, 30])
a.insert(1, 15)
print(a)
```

### Removing Elements — pop()

```python
# pop() removes an element using its index.
import array
a = array.array('i', [10, 20, 30, 40])
a.pop(1)
print(a)
```

### Updating an Array Element

```python
# You can change an element using its index.
import array
a = array.array('i', [10, 20, 30])
a[1] = 50
print(a)
```

### Array Slicing

```python
# Slicing allows you to select a range of elements.
import array
a = array.array('i', [10, 20, 30, 40, 50])
print(a[1:4])
```

### Searching an Element

```python
# Use index() to find the position of an element.
import array
a = array.array('i', [10, 20, 30, 40])
print(a.index(30))
```

### Counting Elements

```python
# Use count() to find how many times a value occurs.
import array
a = array.array('i', [10, 20, 20, 30, 20])
print(a.count(20))
```

### Reversing an Array

```python
import array
a = array.array('i', [10, 20, 30, 40])
a.reverse()
print(a)
```

### Extending an Array

```python
# extend() adds multiple elements to the end.
import array
a = array.array('i', [1, 2, 3])
a.extend([4, 5, 6])
print(a)
```
