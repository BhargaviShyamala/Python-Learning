# Python Strings

```python
# A string is a sequence of characters used to store text.
# Characters can include:
       #Letters → "Python"
       # Numbers → "12345"
       #Symbols → "@#$"
       #Spaces → "Hello World"
# In Python, there is no separate character data type. A single character is also a string.
```

```python
name = "Python"
letter = "P"
print(name)
print(letter)
```

```python
# Creating a String: We can create strings using single quotes or double quotes.
a = 'Hello'
b = "Python"
print(a)
print(b)
```

# Multi-line Strings

```python
# For multiple lines, use triple quotes.
s = """I am learning
Python strings
Today"""
print(s)
```

# Accessing Characters

```python
# Every character in a string has an index.
# Python indexing starts from 0.
s = "PYTHON"
print(s[0])
print(s[3])
print(s[5])
```

# Negative Indexing

```python
# Negative indexing starts from the right side.
s = "PYTHON"
print(s[-1])
print(s[-3])
```

# String Slicing

```python
# Slicing is used to extract a part of a string.
# Syntax: string[start:end]
s = "PYTHON"
print(s[:3])
print(s[3:])
print(s[::-1])
```

# Looping Through a String

```python
# A string is iterable, so we can access every character using a for loop.
s = "PYTHON"
for char in s:
    print(char)
```

# String Immutability

```python
#Strings are immutable.
#Immutable means: Once a string is created, we cannot directly change its characters.
s = "HELLO"
s = "Y" + s[1:]
print(s)
```

# Deleting a String

```python
# We can delete the entire string variable using del.
s = "Hello"
del s
print(s)
```

# Updating a String

```python
# Because strings are immutable, we cannot directly modify them.
s = "Hello World"
s1 = "Y" + s[1:]
print(s1)
```

```python
s = "Python is easy"
s2 = s.replace("easy", "powerful")
print(s2)
```

# Common String Methods

```python
# len(): Returns the number of characters.
s = "Python"
print(len(s))
```

```python
# upper(): Converts the string to uppercase.
s = "hello"
print(s.upper())
```

```python
# lower():Converts the string to lowercase.
s = "PYTHON"
print(s.lower())
```

```python
# strip(): Removes spaces from the beginn
s = "   Python   "
print(s.strip())
```

```python
# replace(): Replaces one part of a string with another.
s = "I like Java"
print(s.replace("Java", "Python"))
```

# String Concatenation

```python
# Concatenation means joining strings.
# We use +.
first = "Hello"
second = "Python"
result = first + " " + second
print(result)
```

# String Repetition

```python
# We can repeat a string using *.
s = "Hi "
print(s * 3)
```

# String Formatting

```python
# Using f-string: f-string is an easy way to put variables inside a string.
name = "Bhargavi"
age = 22
print(f"My name is {name} and I am {age} years old.")
```

# Using format()

```python
name = "Bhargavi"
age = 22
print("My name is {} and I am {} years old.".format(name, age))
```

# String Membership

```python
# The in keyword checks whether something exists inside a string.
s = "Python Programming"
print("Python" in s)
print("Java" in s)
```

# Python Lists

```python
# A list is an ordered, changeable collection of elements written using square brackets [].
numbers = [10, 20, 30, 40]
print(numbers)
```

# Creating a List

```python
# Using Square Brackets []
a = [1, 2, 3]
print(a)
b = ["apple", "banana"]
print(b)
```

```python
data = [10, "Python", 3.5, True]
print(data)
```

```python
# Using list() Constructor: 
a = list((1, 2, 3))
print(a)
b = list("Python")
print(b)
```

```python
# Creating Repeated Elements: The * operator can repeat list elements.
a = [2] * 5
print(a)
```

```python
a = [0] * 4
print(a)
```

```python
# Indexing in Lists: Every element has an index position.
# Python starts indexing from 0.
a = [10, 20, 30, 40]
print(a[0])
print(a[2])
```

```python
# Negative Indexing: python also supports negative indexes.
a = [10, 20, 30, 40]
print(a[-1])
print(a[-2])
```

# Adding Elements

```python
# append(): Adds one element at the end.
a = [1, 2]
a.append(3)
print(a)
```

```python
# insert(): Adds an element at a specific index.
a = [1, 3]
a.insert(1, 2)
print(a)
```

```python
# extend():Adds multiple elements.
a = [1, 2]
a.extend([3, 4, 5])
print(a)
```

```python
# Difference:
   # append()  → adds one element
   #insert()  → adds one element at a specific position
   #extend()  → adds multiple elements
```

# Updating Elements

```python
# Lists are mutable, so we can change their values.
a = [10, 20, 30, 40]
a[1] = 25
print(a)
```

# Removing Elements

```python
# remove():Removes an element by its value.
a = [10, 20, 30]
a.remove(20)
print(a)
```

```python
# pop():Removes an element using its index.
 #If no index is given, it removes the last element.
a = [10, 20, 30]
a.pop()
print(a)
```

```python
# del:del can delete an element using its index.
a = [10, 20, 30]
del a[1]
print(a)
```

```python
# Creating a list
numbers = [10, 20, 30]

print("Original list:", numbers)

# Accessing element
print("First element:", numbers[0])

# Adding element
numbers.append(40)
print("After append:", numbers)

# Inserting element
numbers.insert(1, 15)
print("After insert:", numbers)

# Updating element
numbers[2] = 25
print("After update:", numbers)

# Removing element
numbers.remove(25)
print("After remove:", numbers)

# Removing last element
numbers.pop()
print("After pop:", numbers)
```
