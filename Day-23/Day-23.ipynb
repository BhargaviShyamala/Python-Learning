# Triangular Patterns of Alphabets

### Alphabet Decreasing Triangle

```python
# Print alphabets in a decreasing triangular pattern

n = 5

# Outer loop controls the rows
for i in range(1, n + 1):

    # Inner loop prints alphabets from i to n
    for j in range(i, n + 1):
        print(chr(64 + j), end=" ")

    # Move to the next line
    print()
```

### Same Alphabet Decreasing Triangle

```python
# Print the same alphabet in each row

n = 5

# Outer loop controls the rows
for i in range(n, 0, -1):

    # Print the same alphabet i times
    for j in range(1, i + 1):
        print(chr(64 + i), end=" ")

    # Move to the next line
    print()
```

### Increasing Alphabet Triangle

```python
# Print alphabets in an increasing triangular pattern

n = 5

# Outer loop controls the rows
for i in range(1, n + 1):

    # Print alphabets from A up to the current row
    for j in range(1, i + 1):
        print(chr(64 + j), end=" ")

    # Move to the next line
    print()
```

###  Reverse Starting Alphabet Triangle

```python

# Print alphabets starting from different letters

n = 5

# Outer loop starts from E and goes to A
for i in range(n, 0, -1):

    # Print alphabets from the current letter to E
    for j in range(i, n + 1):
        print(chr(64 + j), end=" ")

    # Move to the next line
    print()
```

###  Same Alphabet Decreasing Count

```python
# Print the same alphabet with decreasing number of elements

n = 5

# Outer loop controls the alphabet
for i in range(1, n + 1):

    # Print the current alphabet n-i+1 times
    for j in range(n, i - 1, -1):
        print(chr(64 + i), end=" ")

    # Move to the next line
    print()
```

# Swastika Pattern in Python

```python
# Program to print a Swastika pattern

def swastika(row, col):

    # Loop through each row
    for i in range(row):

        # Loop through each column
        for j in range(col):

            # Upper half of the pattern
            if i < row // 2:

                # Left side
                if j < col // 2:
                    if j == 0:
                        print("*", end="")
                    else:
                        print(" ", end=" ")

                # Middle column
                elif j == col // 2:
                    print("*", end=" ")

                # Right side
                else:
                    if i == 0:
                        print("*", end=" ")
                    else:
                        print(" ", end=" ")

            # Middle row
            elif i == row // 2:
                print("*", end=" ")

            # Lower half of the pattern
            else:

                # Middle column and last column
                if j == col // 2 or j == col - 1:
                    print("*", end=" ")

                # Last row
                elif i == row - 1:

                    # Print stars on the left side
                    if j <= col // 2:
                        print("*", end=" ")
                    else:
                        print(" ", end=" ")

                # Other positions
                else:
                    print(" ", end=" ")

        # Move to the next line
        print()


# Number of rows and columns
# Both should be odd and equal
row = 7
col = 7

# Call the function
swastika(row, col)
```
