# Polymorphism in Python

```python
# Polymorphism = Many Forms
# In Python, polymorphism means: The same method, function, or operator can behave differently depending on the object or data.
#Simple real-life example
   # Suppose we have:
       # Dog → sound() → Bark
       # Cat → sound() → Meow
       # Cow → sound() → Moo
#The method name is the same: sound()
# But the behavior is different.
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
class Cat(Animal):
    def sound(self):
        print("Cat meows")
dog = Dog()
cat = Cat()
dog.sound()
cat.sound()
```

### Polymorphism with Multiple Child Classes

```python
class Animal:
    def sound(self):
        print("Animal sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
class Cow(Animal):
    def sound(self):
        print("Moo")
animals = [Dog(), Cat(), Cow()]
for animal in animals:
    animal.sound()
```

### Polymorphism using Functions

```python
# A function can accept different types of objects.
class Dog:
    def sound(self):
        return "Bark"
class Cat:
    def sound(self):
        return "Meow"
def make_sound(animal):
    print(animal.sound())
dog = Dog()
cat = Cat()
make_sound(dog)
make_sound(cat)
```

### Duck Typing

```python
# Python follows an important idea called Duck Typing.
#"If an object behaves like the required object, Python can use it."
class Dog:
    def sound(self):
        print("Bark")
class Cat:
    def sound(self):
        print("Meow")
class Cow:
    def sound(self):
        print("Moo")
def make_sound(animal):
    animal.sound()
make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
```

### Polymorphism with len()

```python
# The + operator behaves differently depending on the data type.
print(10 + 20) # Integers
print("Hello " + "Python")# Strings
```

```python
# Python's built-in functions are also polymorphic.
print(len("Python"))
print(len([10, 20, 30]))
print(len((1, 2, 3, 4)))
```

### Polymorphism with + Operator

```python
print(10 + 20)
print("Hello " + "Python")
```

### Polymorphism with *

```python
# The * operator also behaves differently.
print(5 * 3)
print("Hi " * 3)
print([1, 2] * 2)
```

### Polymorphism with Different Classes

```python
class CreditCard:
    def pay(self):
        print("Payment using Credit Card")
class UPI:
    def pay(self):
        print("Payment using UPI")
class Wallet:
    def pay(self):
        print("Payment using Wallet")
def make_payment(payment_method):
    payment_method.pay()
make_payment(CreditCard())
make_payment(UPI())
make_payment(Wallet())
```

### Compile-Time Polymorphism / Method Overloading

```python
# Python does not support traditional method overloading like Java or C++.
# For example, this does not work as traditional overloading:
class Calculator:
    def add(self, a, b):
        return a + b
    def add(self, a, b, c):
        return a + b + c
```

```python
class Calculator:
    def add(self, a, b, c=0):
        return a + b + c
calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

### Method Overloading Using Default Arguments

```python
class Calculator:

    def add(self, a=0, b=0, c=0):
        return a + b + c


calc = Calculator()

print(calc.add())
print(calc.add(10))
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

### Using *args

```python
class Calculator:
    def add(self, *numbers):
        total = 0
        for number in numbers:
            total += number
        return total
calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(1, 2, 3, 4, 5))
```

### Real-Life Employee Example

```python
class Employee:
    def work(self):
        print("Employee is working")
class Developer(Employee):
    def work(self):
        print("Developer writes code")
class Tester(Employee):
    def work(self):
        print("Tester tests software")
class Manager(Employee):
    def work(self):
        print("Manager manages the team")
employees = [
    Developer(),
    Tester(),
    Manager()
]
for employee in employees:
    employee.work()
```

### Vehicles Example

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")
class Car(Vehicle):
    def start(self):
        print("Car starts with a key")
class Bike(Vehicle):
    def start(self):
        print("Bike starts with a button")
class Bus(Vehicle):
    def start(self):
        print("Bus starts with a key")
vehicles = [Car(), Bike(), Bus()]
for vehicle in vehicles:
    vehicle.start()
```
