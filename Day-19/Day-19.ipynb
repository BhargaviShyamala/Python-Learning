##### Sum of Digits of a Number

```python
# Using while loop
def sumOfDigits(n):
    total = 0

    while n != 0:
        last = n % 10
        total = total + last
        n = n // 10

    return total


n = 12345
print(sumOfDigits(n))
```

```python
# Using recursion
def sumOfDigits(n):
    if n == 0:
        return 0
    return (n % 10) + sumOfDigits(n // 10)

n = 1234567
print(sumOfDigits(n))
```

```python
# Using String Conversion
def sumOfDigits(n):
    total = 0
    for digit in str(n):
        total = total + int(digit)
    return total
n = 12345
print(sumOfDigits(n))
```

##### Reverse Digits of a Number

```python
# Using while loop
def reverseDigits(n):
    rev = 0
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10
    return rev
n = 12345
print(reverseDigits(n))
```

```python
def reverseDigits(n):
    rev = 0
    while n > 0:
        digit = n % 10
        rev = rev * 10 + digit
        n = n // 10
    return rev
n = 200
print(reverseDigits(n))
```

```python
# Using String Slicing
def reverseDigits(n):
    s = str(n)
    s = s[::-1]
    return int(s)
n = 12345
print(reverseDigits(n))
```

##### Program for Prime Number Check

```python
# A prime number is a number greater than 1 that can be divided exactly only by: 1 and itself.
def isPrime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False

    return True
n = 7
if isPrime(n):
    print("true")
else:
    print("false")
```

```python
def isPower(x, y):
    if x == 1:
        return y == 1
    power = 1
    while power < y:
        power = power * x
    return power == y
x = 10
y = 1001
print(isPower(x, y))
```
