## Diagonal Star Pattern

```python
# Program to print a diagonal star pattern

def pattern(n):

    # Loop through each row
    for i in range(n):

        # Loop through each column
        for j in range(n):

            # Print star on:
            # first row, first column,
            # main diagonal, last row,
            # last column and secondary diagonal
            if (i == 0 or
                j == 0 or
                i == j or
                i == n - 1 or
                j == n - 1 or
                i + j == n - 1):

                print("*", end="")

            else:
                # Print space
                print(" ", end="")

        # Move to the next line
        print()


# Size of the pattern
# Use an odd number
n = 7

# Call the function
pattern(n)
```

## Diamond Diagonal Star Pattern

```python
# Program to print a diamond diagonal star pattern

def pattern(n):

    # Calculate the center position
    center = (n - 1) // 2

    # Loop through each row
    for i in range(n):

        # Loop through each column
        for j in range(n):

            # Check the diagonal and center conditions
            if (i + j == center or
                i - j == center or
                j - i == center or
                i + j == 3 * n // 2 - 1 or
                i == center or
                j == center):

                print("*", end=" ")

            else:
                # Print spaces
                print(" ", end=" ")

        # Move to the next line
        print()


# Size of the pattern
# Use an odd number
n = 9

# Call the function
pattern(n)
```

### Double Triangle Star Pattern

```python
# Program to print a double triangle star pattern

def pattern(n):

    # Upper half of the pattern
    for i in range(1, n + 1):

        # Loop through columns
        for j in range(1, 2 * n):

            # Print stars on the left side
            if i > n - j + 1:
                print(" ", end=" ")
            else:
                print("*", end=" ")

            # Print stars on the right side
            if i + n - 1 > j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

        # Move to the next line
        print()

    # Lower half of the pattern
    for i in range(1, n + 1):

        # Loop through columns
        for j in range(1, 2 * n):

            # Print stars on the left side
            if i < j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

            # Print stars on the right side
            if i < 2 * n - j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

        # Move to the next line
        print()


# Call the function
pattern(5)
```

### Increasing and Decreasing Double Triangle


```python
# Program to print an increasing and decreasing star pattern

def pattern(n):

    # Upper half of the pattern
    for i in range(1, n + 1):

        # Loop through columns
        for j in range(1, 2 * n + 1):

            # Print stars on the left side
            if i < j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

            # Print stars on the right side
            if i <= 2 * n - j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

        # Move to the next line
        print()

    # Lower half of the pattern
    for i in range(1, n + 1):

        # Loop through columns
        for j in range(1, 2 * n + 1):

            # Print stars on the left side
            if i > n - j + 1:
                print(" ", end=" ")
            else:
                print("*", end=" ")

            # Print stars on the right side
            if i + n > j:
                print(" ", end=" ")
            else:
                print("*", end=" ")

        # Move to the next line
        print()


# Call the function
pattern(5)
```

### Diagonal Slash Pattern

```python
# Program to print a diagonal slash pattern

def pattern(n):

    # Loop through rows
    for i in range(1, n + 1):

        # Loop through columns
        for j in range(1, n + 1):

            # Check for diagonals
            if i == j or i + j == n + 1:

                # Print forward slash
                if i + j == n + 1:
                    print("/", end="")

                # Print backslash
                else:
                    print("\\", end="")

            else:
                # Print star
                print("*", end="")

        # Move to the next line
        print()


# Call the function
pattern(8)
```

### Decreasing Number Pattern

```python
# Program to print a decreasing number pattern

def pattern(n):

    # Loop through rows
    for i in range(1, n + 1):

        # Print numbers in decreasing order
        for j in range(n - i, -1, -1):
            print(j, end=" ")

        # Move to the next line
        print()


# Call the function
pattern(8)
```

### Triangle Number Pattern

```python
# Program to print a special number triangle

def pattern(n):

    # Starting number
    k = 1

    # Loop through rows
    for i in range(1, n + 1):

        # Store the starting number of the row
        p = k

        # Loop through columns
        for j in range(1, i + 1):

            # Print the current number
            print(p, end=" ")

            # Calculate the next number
            p = p - (n - i + j)

        # Move to the next line
        print()

        # Calculate starting number for next row
        k = k + 1 + (n - i)


# Call the function
pattern(7)
```
