# Python Sets 


```python
# A set in Python is a collection used to store unique values.
# Main features of a Set:
    # 1. Duplicates are not allowed
    # 2. Unordered — elements do not have fixed positions.
    #3. No indexing — you cannot use s[0].
    #4. Mutable — you can add or remove elements.
    #5. Can contain different data types.
    #6. Supports operations like union, intersection, difference, etc.
```

### Creating a Set

```python
s = {10, 50, 20}
print(s)
print(type(s))
# Important: The order of elements is not guaranteed.
```

### Duplicate Values

```python
# A set automatically removes duplicate values.
s = {"Geeks", "for", "Geeks"}
print(s)
```

### Sets Do Not Support Indexing

```python
# You cannot access a set using an index.
s = {10, 20, 30}
print(s[0])
```

```python
numbers = [10, 20, 30]
print(numbers[0])
```

### Sets Are Mutable


```python
# A set itself can be changed.
# You can add or remove elements.
s = {10, 20, 30}
s.add(40)
print(s)
```

### Creating a Set Using set()

```python
# The set() function can convert another collection into a set.
numbers = [10, 20, 10, 30, 20]
s = set(numbers)
print(s)
```

```python
names = ["A", "B", "A", "C", "B"]
unique_names = set(names)
print(unique_names)
```

### Heterogeneous Set

```python
# A set can contain different data types.
s = {"Python", 10, 25.5, True}
print(s)
```

### Adding Elements — add()


```python
# The add() method adds one element.
s = {"a", "b", "c"}
s.add("d")
print(s)
```

### Union

```python
# Union combines two sets and removes duplicates.
a = {1, 2, 3}
b = {3, 4, 5}
result = a.union(b)
print(result)
```

### Intersection

```python
# Intersection gives the elements that are present in both sets.
a = {1, 2, 3}
b = {2, 3, 4}
result = a.intersection(b)
print(result)
```

### Difference

```python
# Difference gives elements that are present in the first set but not in the second set.
a = {1, 2, 3}
b = {2, 3, 4}
result = a.difference(b)
print(result)
```

### Clear

```python
# clear() removes all elements from a set.
s = {10, 20, 30}
s.clear()
print(s)
```

### Membership Operators

```python
# You can check whether an element exists in a set.
s = {10, 20, 30}
print(20 in s)
print(50 not in s)
```

### Subset

```python
# A set is a subset if all its elements are present in another set.
a = {1, 2}
b = {1, 2, 3, 4}
print(a.issubset(b))
print(a <= b)
```

### Superset

```python
# A set is a superset if it contains all elements of another set.
a = {1, 2, 3, 4}
b = {1, 2}
print(a.issuperset(b))
print(a >= b)
```

### Frozenset

```python
# A frozenset is an immutable version of a set.
s = frozenset([10, 20, 30])
print(s)
```

```python
# You cannot add or remove elements:
s = frozenset([10, 20, 30])
print(s)
s.add(40)
```

### Symmetric Difference

```python
# This gives elements that are in either set, but not in both.
a = {1, 2, 3}
b = {3, 4, 5}
result = a ^ b
print(result)
```
