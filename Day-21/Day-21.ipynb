### Decimal to Binary Conversion

```python
# Program: Convert Decimal Number to Binary Number

def decToBinary(n):
    # Create an empty string to store the binary value
    binary = ""

    # If the number is 0, return "0"
    if n == 0:
        return "0"

    # Repeat until the number becomes 0
    while n > 0:
        # Find the remainder when divided by 2
        remainder = n % 2

        # Add the remainder to the beginning of the binary string
        binary = str(remainder) + binary

        # Divide the number by 2 using integer division
        n = n // 2

    # Return the binary number
    return binary


# Take decimal number as input from the user
n = int(input("Enter a number: "))

# Call the function and display the binary result
print("Binary:", decToBinary(n))
```

## Find N-th Term of Series: 1, 3, 6, 10, 15, 21...

```python
# Function to find the sum of first n natural numbers
def term(n):
    # Use the formula: n * (n + 1) / 2
    return n * (n + 1) // 2


# Take n as input from the user
n = int(input("Enter n: "))

# Call the function and print the result
print(term(n))
```

## Armstrong Number

```python
# An Armstrong number is a number where the sum of each digit raised to the power of the number of digits is equal to the original number.
# Function to check whether a number is an Armstrong number
def armstrong(n):
    # Convert the number into a string to access each digit
    number = str(n)

    # Find the number of digits
    digits = len(number)

    # Initialize the sum as 0
    total = 0

    # Loop through each digit in the number
    for digit in number:
        # Raise the digit to the power of the number of digits
        # and add it to the total
        total += int(digit) ** digits

    # Check whether the calculated total is equal to the original number
    return total == n


# Take a number as input from the user
n = int(input("Enter a number: "))

# Check whether the number is an Armstrong number
if armstrong(n):
    # Print true if it is an Armstrong number
    print("true")
else:
    # Print false if it is not an Armstrong number
    print("false")
```

## Palindrome Number

```python
# A palindrome number is a number that reads the same from left to right and right to left.
# Function to check whether a number is a palindrome
def isPalindrome(n):
    # Convert negative numbers to positive using abs()
    original = abs(n)

    # Initialize reverse number as 0
    reverse = 0

    # Store the original number in a temporary variable
    temp = original

    # Repeat until all digits are processed
    while temp > 0:
        # Get the last digit of the number
        digit = temp % 10

        # Add the digit to the reverse number
        reverse = reverse * 10 + digit

        # Remove the last digit from the temporary number
        temp = temp // 10

    # Check whether the reversed number is equal to the original
    return reverse == original


# Take a number as input from the user
n = int(input("Enter a number: "))

# Check whether the number is a palindrome
if isPalindrome(n):
    # Print True if the number is a palindrome
    print("True")
else:
    # Print False if the number is not a palindrome
    print("False")
```
