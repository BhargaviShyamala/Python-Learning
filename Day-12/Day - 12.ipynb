# Inheritance in Python

```python
# Inheritance is an OOP concept where a child class gets the properties and methods of a parent class.
# Think of it like:
     # Parent → Child
     # Animal → Dog
# The Dog can use methods defined inside Animal without writing them again.
```

```python
# Basic Inheritance:
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal name:", self.name)
class Dog(Animal):
    def sound(self):
        print(self.name, "barks")
d = Dog("Buddy")
d.info()
d.sound()
```

### super() 

```python
#  in Inheritance: super() is used when we want to access functionality from the parent class inside the child class.
class Animal:
    def __init__(self, name):
        self.name = name
    def info(self):
        print("Animal name:", self.name)
class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
    def details(self):
        print(self.name, "is a", self.breed)
d = Dog("Buddy", "Golden Retriever")
d.info()
d.details()
```

###  Method Overriding

```python
# Method overriding means the child class creates a method with the same name as a method in the parent class.
class Animal:
    def sound(self):
        print("Animal sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
d = Dog()
d.sound()
```

### Inheritance with Constructor

```python
class Person:
    def __init__(self, name):
        self.name = name
    def display_name(self):
        print("Name:", self.name)
class Student(Person):
    def display_student(self):
        print("Student:", self.name)
s = Student("Bhargavi")
s.display_name()
s.display_student()
```

### Child Has Its Own Constructor

```python
class Person:
    def __init__(self, name):
        self.name = name
class Student(Person):
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
s = Student("Bhargavi", 85)
s.display()
```

## Employee Example

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
class Developer(Employee):
    def show_role(self):
        print("Role: Developer")
d = Developer("Bhargavi", 50000)
d.display()
d.show_role()
```

### Student → College Example

```python
class College:
    def college_name(self):
        print("College: JNTUH")
class Student(College):
    def student_name(self):
        print("Student: Bhargavi")
s = Student()
s.college_name()
s.student_name()
```

### Vehicle → Car

```python
class Vehicle:
    def start(self):
        print("Vehicle starts")
class Car(Vehicle):
    def drive(self):
        print("Car is driving")
c = Car()
c.start()
c.drive()
```

### Parent + Child Attributes

```python
class Person:
    def __init__(self, name):
        self.name = name
class Student(Person):
    def __init__(self, name, roll_no):
        super().__init__(name)
        self.roll_no = roll_no
    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
s = Student("Priya", 101)
s.display()
```

### Multilevel Inheritance

```python
class Grandparent:
    def house(self):
        print("Grandparent has a house")
class Parent(Grandparent):
    def car(self):
        print("Parent has a car")
class Child(Parent):
    def bike(self):
        print("Child has a bike")
c = Child()
c.house()
c.car()
c.bike()
```

### Multiple Inheritance

```python
# Python also supports multiple inheritance.
# One child can inherit from two parent classes.
class Father:
    def father_property(self):
        print("Father's property")
class Mother:
    def mother_property(self):
        print("Mother's property")
class Child(Father, Mother):
    def child_property(self):
        print("Child's property")
c = Child()
c.father_property()
c.mother_property()
c.child_property()
```

### Multiple Inheritance with super()

```python
class A:
    def show(self):
        print("A")
class B(A):
    def show(self):
        print("B")
        super().show()
obj = B()
obj.show()
```

### Real-World Banking Example

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    def display_balance(self):
        print("Account Holder:", self.name)
        print("Balance:", self.balance)
class SavingsAccount(BankAccount):
    def add_interest(self):
        interest = self.balance * 0.05
        self.balance += interest
    def display(self):
        print("Updated Balance:", self.balance)
account = SavingsAccount("Bhargavi", 10000)
account.display_balance()
account.add_interest()
account.display()
```

### Inheritance + Method Overriding + super()

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)
class Developer(Employee):
    def __init__(self, name, salary, language):
        super().__init__(name, salary)
        self.language = language
    def display(self):
        super().display()
        print("Programming Language:", self.language)
d = Developer("Bhargavi", 60000, "Python")
d.display()
```
