# Abstraction in Python

```python
# Abstraction means hiding unnecessary implementation details and showing only the important functionality to the user.
# Simple real-life example
#When you use an ATM:
       #You enter your PIN.
       #you select Withdraw.
       #You enter the amount.
       #you receive money.
#You don't need to know:
      #How the bank verifies your PIN.
      #How the ATM communicates with the bank server.
      #How the transaction is processed internally.
#So:
     #What you see = essential functionality
     #What is hidden = implementation details
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
```

### Implementing the Abstract Method

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
class Dog(Animal):
    def make_sound(self):
        return "Bark"
dog = Dog()
print(dog.make_sound())
```

### Multiple Child Classes

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
class Dog(Animal):
    def make_sound(self):
        return "Bark"
class Cat(Animal):
    def make_sound(self):
        return "Meow"
class Cow(Animal):
    def make_sound(self):
        return "Moo"
dog = Dog()
cat = Cat()
cow = Cow()
print(dog.make_sound())
print(cat.make_sound())
print(cow.make_sound())
```


### Abstract Class with Concrete Method

```python
# An abstract class can contain both:
   # Abstract methods
   # Concrete methods
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
    def move(self):
        return "Animal is moving"
class Dog(Animal):
    def make_sound(self):
        return "Bark"
dog = Dog()
print(dog.make_sound())
print(dog.move())
```

### Real-Life Example: Payment System

```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
class CreditCard(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using Credit Card"
class UPI(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using UPI"
class PayPal(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using PayPal"
p1 = CreditCard()
p2 = UPI()
p3 = PayPal()
print(p1.pay(1000))
print(p2.pay(500))
print(p3.pay(1500))
```

### Abstract Property

```python
# We can also create an abstract property.
from abc import ABC, abstractmethod
class Animal(ABC):
    @property
    @abstractmethod
    def species(self):
        pass
class Dog(Animal):
    @property
    def species(self):
        return "Canine"
dog = Dog()
print(dog.species)
```

### Simple Shape

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Circle(Shape):
    def area(self):
        return 3.14 * 5 * 5
class Rectangle(Shape):
    def area(self):
        return 10 * 5
c = Circle()
r = Rectangle()
print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

### Employee Salary

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    @abstractmethod
    def calculate_salary(self):
        pass
class FullTimeEmployee(Employee):
    def calculate_salary(self):
        return 50000
class PartTimeEmployee(Employee):
    def calculate_salary(self):
        return 25000
e1 = FullTimeEmployee()
e2 = PartTimeEmployee()
print("Full Time Salary:", e1.calculate_salary())
print("Part Time Salary:", e2.calculate_salary())
```

### Bank Account

```python
from abc import ABC, abstractmethod
class Bank(ABC):
    @abstractmethod
    def interest_rate(self):
        pass
class SBI(Bank):
    def interest_rate(self):
        return 7
class HDFC(Bank):
    def interest_rate(self):
        return 7.5
sbi = SBI()
hdfc = HDFC()
print("SBI Interest:", sbi.interest_rate(), "%")
print("HDFC Interest:", hdfc.interest_rate(), "%")
```

### Payment System

```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass
class UPI(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI")
class CreditCard(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Credit Card")
class NetBanking(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Net Banking")
upi = UPI()
card = CreditCard()
net = NetBanking()
upi.pay(1000)
card.pay(2000)
net.pay(3000)
```

### Abstract Class + Normal Method

```python
# An abstract class can also contain a normal/concrete method.
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    def eat(self):
        print("Animal is eating")
class Dog(Animal):
    def sound(self):
        print("Dog says Bark")
dog = Dog()
dog.sound()
dog.eat()
```

### Multiple Abstract Methods

```python
from abc import ABC, abstractmethod
class Student(ABC):
    @abstractmethod
    def study(self):
        pass
    @abstractmethod
    def attend_class(self):
        pass
class CollegeStudent(Student):
    def study(self):
        print("Student is studying")
    def attend_class(self):
        print("Student is attending class")
student = CollegeStudent()
student.study()
student.attend_class()
```

### Food Delivery Example

```python
from abc import ABC, abstractmethod
class FoodDelivery(ABC):
    @abstractmethod
    def deliver(self):
        pass
class Swiggy(FoodDelivery):
    def deliver(self):
        print("Food delivered by Swiggy")
class Zomato(FoodDelivery):
    def deliver(self):
        print("Food delivered by Zomato")
s = Swiggy()
z = Zomato()
s.deliver()
z.deliver()
```

### Database

```python
from abc import ABC, abstractmethod

class Database(ABC):

    @abstractmethod
    def connect(self):
        pass

    @abstractmethod
    def fetch_data(self):
        pass
class MySQL(Database):
    def connect(self):
        print("Connected to MySQL")

    def fetch_data(self):
        print("Fetching data from MySQL")
class MongoDB(Database):
    def connect(self):
        print("Connected to MongoDB")
    def fetch_data(self):
        print("Fetching data from MongoDB")
mysql = MySQL()
mongo = MongoDB()
mysql.connect()
mysql.fetch_data()
mongo.connect()
mongo.fetch_data()
```
