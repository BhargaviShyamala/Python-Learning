#  Python Operators

```python
# An operator is a special symbol or keyword used to perform an operation on values or variables.
```

```python
a = 10
b = 5
print(a + b)
```

# Arithmetic Operators

```python
# Arithmetic operators are special symbols used to perform mathematical operations .
#such as addition, subtraction, multiplication, and division on values or variables.
a = 15
b = 4

# Addition: adds two numbers
print("Addition:", a + b)

# Subtraction: subtracts b from a
print("Subtraction:", a - b)

# Multiplication: multiplies two numbers
print("Multiplication:", a * b)

# Division: divides a by b and returns a decimal value
print("Division:", a / b)

# Floor Division: divides and returns the whole number part
print("Floor Division:", a // b)

# Modulus: returns the remainder after division
print("Modulus:", a % b)

# Exponentiation: raises a to the power of b
print("Exponentiation:", a ** b)
```

# Comparison Operators

```python
#Comparison operators are used to compare two values or variables.
#They return either True or False based on the comparison.
a = 13
b = 33

# Greater than: checks if a is greater than b
print("Greater than:", a > b)

# Less than: checks if a is less than b
print("Less than:", a < b)

# Equal to: checks if a and b have the same value
print("Equal to:", a == b)

# Not equal to: checks if a and b have different values
print("Not equal to:", a != b)

# Greater than or equal to: checks if a is greater than or equal to b
print("Greater than or equal to:", a >= b)

# Less than or equal to: checks if a is less than or equal to b
print("Less than or equal to:", a <= b)
```

# Logical Operators

```python
# Logical operators are used to combine two or more conditions and return a result as True or False.
a = 10
b = 20

# AND operator
print(a < b and b > 15)

# OR operator
print(a > b or b > 15)

# NOT operator
print(not(a < b))
```

# Bitwise Operators

```python
# Bitwise operators are used to perform operations on the binary (0 and 1) representation of integers.
a = 5
b = 3

# Bitwise AND: returns 1 if both bits are 1
print("Bitwise AND:", a & b)

# Bitwise OR: returns 1 if at least one bit is 1
print("Bitwise OR:", a | b)

# Bitwise XOR: returns 1 if the bits are different
print("Bitwise XOR:", a ^ b)

# Bitwise NOT: reverses the bits
print("Bitwise NOT:", ~a)

# Left Shift: shifts the bits to the left by 1 position
print("Left Shift:", a << 1)

# Right Shift: shifts the bits to the right by 1 position
print("Right Shift:", a >> 1)
```

# Assignment Operators

```python
# Assignment operators are used to assign or update values.
a = 10
a += 5
print(a)
```

# Identity Operators

```python
#There are two identity operators: is and is not
a = [1, 2, 3]
b = a
print(a is b)
```

```python
a = [1, 2]
b = [1, 2]
print(a == b)
print(a is b)
```

# Membership Operators

```python
# Membership operators check whether a value exists inside a sequence such as a list, tuple, string, or set.
# Two operators: in and not in
my_list = [10, 20, 30, 40]
print(20 in my_list)
print(50 in my_list)
print(50 not in my_list)
```

# Ternary Operator

```python
# A ternary operator allows us to write a simple if-else in one line.
age = 20
result = "Adult" if age >= 18 else "Minor"
print(result)
```

# Operator Precedence

```python
# Operator precedence tells Python which operation should be performed first when an expression contains multiple operators.
a = 10          # Assign 10 to variable a
b = 5           # Assign 5 to variable b
c = 2           # Assign 2 to variable c
result = a + b * c   # Multiplication is performed first, then addition
print("Result:", result)   # Display the final result
```

# Operator Associativity

```python
# Operator associativity tells Python which operation to perform first when two or more operators have the same precedence.
a = 10              # Assign 10 to variable a
b = 5               # Assign 5 to variable b
c = 2               # Assign 2 to variable c
result = a - b + c  # + and - have same precedence, so evaluation is from left to right
print("Result:", result)  # Display the final result
```
