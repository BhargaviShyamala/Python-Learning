# Nested Loops

```python
# A nested loop means having one loop inside another loop.
for i in range(1, 5):
    for j in range(i):
        print(i, end=" ")
    print()
```

# Functions

```python
# A function is a reusable block of code that performs a specific task.
# Instead of writing the same code again and again, we write it once inside a function and call it whenever needed.
# Why do we use functions?
    #Functions help us to:
       #Avoid repeating code
       #Make programs shorter and cleaner
       # Organize code into small parts
       # Reuse the same logic
       # Make programs easier to understand and maintain
```

# Defining a Function

```python
# We use the def keyword to create a function.
def welcome():
    print("Welcome to Python")
welcome()
```

# Function with Arguments

```python
# Arguments allow us to send data into a function.
def add(a, b):
    print(a + b)
add(10, 20)
```

# Function with Return

```python
# return sends a result back to the place where the function was called.
def add(a, b):
    return a + b
result = add(10, 20)
print(result)
```

# Default Arguments

```python
# A default argument already has a value.
def student(name, age=20):
    print("Name:", name)
    print("Age:", age)
student("Ravi")
```

# Keyword Arguments

```python
# We can pass values using parameter names.
def student(name, age):
    print("Name:", name)
    print("Age:", age)
student(age=22, name="Ravi")
```

# Positional Arguments

```python
# Values are assigned according to their position/order.
def student(name, age):
    print("Name:", name)
    print("Age:", age)
student("Ravi", 22)
```

# Arbitrary Arguments — *args

```python
# *args allows a function to accept multiple positional arguments.
def numbers(*args):
    for num in args:
        print(num)
numbers(10, 20, 30, 40)
```

# Arbitrary Keyword Arguments — **kwargs

```python
# **kwargs allows a function to accept multiple keyword arguments.
def student(**kwargs):
    print(kwargs)
student(name="Ravi", age=22, city="Hyderabad")
```

# Nested Function

```python
# A function inside another function is called a nested function.
def outer():
    print("Outer function")
    def inner():
        print("Inner function")
    inner()
outer()
```

# Return Statement

```python
# return sends a value back to the caller.
def square(num):
    return num * num
result = square(5)
print(result)
```

# Returning Multiple Values

```python
# A function can return multiple values.
def calculate(a, b):
    return a + b, a - b
addition, subtraction = calculate(10, 5)
print("Addition:", addition)
print("Subtraction:", subtraction)
```

# Mutable and Immutable Objects

```python
# Mutable objects: These can be changed after creation.
def change_list(numbers):
    numbers[0] = 100
numbers = [10, 20, 30]
change_list(numbers)
print(numbers)
```

```python
# Immutable objects: These cannot be changed after creation.
def change_number(x):
    x = 100
number = 10
change_number(number)
print(number)
```
