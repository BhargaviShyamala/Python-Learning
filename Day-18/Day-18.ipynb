#####  Check Even or Odd

```python
def isEven(n):
    if n % 2 == 0:
        return True
    else:
        return False
n = 15
if isEven(n):
    print("true")
else:
    print("false")
```

#####  Multiplication Table 


```python
n = int(input("Enter a number: "))
for i in range(1, 11):
    print(n, "*", i, "=", n * i)
```

```python
def printTable(n):
    for i in range(1, 11):
        print(f"{n} * {i} = {n * i}")
n = int(input("Enter a number: "))
printTable(n)
```

##### Swap Two Numbers

```python
# Using a Third Variable
a = int(input("Enter a: "))
b = int(input("Enter b: "))
temp = a
a = b
b = temp
print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
# Without Third Variable
a = int(input("Enter a: "))
b = int(input("Enter b: "))
a, b = b, a
print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
# Using Arithmetic Operators
a = int(input("Enter a: "))
b = int(input("Enter b: "))
a = a + b
b = a - b
a = a - b
print("After swapping:")
print("a =", a)
print("b =", b)
```

```python
# Using XOR
a = int(input("Enter a: "))
b = int(input("Enter b: "))
a = a ^ b
b = a ^ b
a = a ^ b
print("After swapping:")
print("a =", a)
print("b =", b)
```

##### Closest to n and Divisible by m

```python
# Using Loop
n = int(input("Enter n: "))
m = int(input("Enter m: "))
closest = 0
min_difference = float('inf')
for i in range(n - abs(m), n + abs(m) + 1):
    if i % m == 0:
        difference = abs(n - i)
        if difference < min_difference:
            closest = i
            min_difference = difference
        elif difference == min_difference:
            if abs(i) > abs(closest):
                closest = i
print("Closest number =", closest)
```

##### dice problem

```python
n = int(input("Enter the dice number: "))
if n == 1:
    opposite = 6
elif n == 2:
    opposite = 5
elif n == 3:
    opposite = 4
elif n == 4:
    opposite = 3
elif n == 5:
    opposite = 2
else:
    opposite = 1
print("Opposite face =", opposite)
```
