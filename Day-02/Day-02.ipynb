# Input and Output in Python


```python
In Python:
# Input → taking information from the user.
# Output → showing information to the user.
```

```python
# input()-Taking Input
name = input("Enter your name: ")
print("Hello", name)
```

```python
age = input("Enter age: ")
```

```python
# Printing Multiple Values
name = "Ravi"
age = 25
city = "Hyderabad"

print(name, age, city)
```

```python
# Taking Multiple Inputs
x, y = input("Enter two numbers: ").split() # split() separates the input into different parts.
print(x, y)
```

```python
# Taking Integer Input
age = int(input("Enter your age: "))
print(age)
```

```python
#Taking Float Input
marks = float(input("Enter your marks: "))
print(marks)
```

# Python Variables

```python
# A variable is a name used to store a value.
# Think of it like a box with a name
x = 5
name = "Alex"

print(x)
print(name)
```

```python
# Rules for Naming Variables
    # Valid names:
           age = 21
           _colour = "lilac"
           total_score = 90
           student1 = "Ravi"
    # Invalid names:
         1name = "Alex"       # Cannot start with a digit
         class = 10          # Keyword cannot be used
         user-name = "Doe"   # Hyphen is not allowed
# Important rules:
1. Can contain letters, numbers and _
2. Cannot start with a number
3. Python is case-sensitive
4. Cannot use Python keywords
```

```python
age = 20
Age = 30
print(age)
print(Age)
```

# Different Ways of Assignment:

```python
#Basic assignment:
x = 5
y = 3.14
name = "Python"
```

```python
# Same value to multiple variables
a = b = c = 100
print(a, b, c)
#Different values to different variables
x, y, z = 1, 2.5, "Python"
print(x, y, z)
```

```python
# Object Reference
x = "Geeks"
y = x
print(x)
print(y)
```

```python
# Variable Reassignment
x = 1
y = x
y = y + 1
print(x)
print(y)
```

```python
# Deleting a Variable
x = 10
del x
print(x)
```

```python
# Swapping Variables
a = 5
b = 10
a, b = b, a
print(a)
print(b)
```
