# Perfect Number

```python
# A perfect number is a number whose proper divisors add up to the number itself.
def isPerfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total += i

    return total == n


n = int(input("Enter a number: "))

if isPerfect(n):
    print("true")
else:
    print("false")
# The range(1, n) means we check from 1 to n-1, because the number itself is not considered a proper divisor.
```

# Program to add two fractions

```python
from math import gcd

def addFraction(a, b):
    num = a[0] * b[1] + b[0] * a[1]
    den = a[1] * b[1]

    common = gcd(num, den)

    num = num // common
    den = den // common

    return [num, den]


a = [1, 2]
b = [3, 2]

ans = addFraction(a, b)

print(ans)
```

# Find Day of the Week

```python
# This program finds which day of the week a given date falls on.
def getDayOfWeek(date):
    d, m, y = date

    t = [0, 3, 2, 5, 0, 3, 5, 1, 4, 6, 2, 4]

    if m < 3:
        y -= 1

    day = (y + y // 4 - y // 100 + y // 400 + t[m - 1] + d) % 7

    weekDays = [
        "Sunday",
        "Monday",
        "Tuesday",
        "Wednesday",
        "Thursday",
        "Friday",
        "Saturday"
    ]

    return weekDays[day]


date = [30, 8, 2010]

print(getDayOfWeek(date))
```

# Nth Fibonacci Number

```python
# The Fibonacci series is: 0, 1, 1, 2, 3, 5, 8, 13, 21, ...
# Each number is obtained by adding the previous two numbers.
def nthFibonacci(n):
    if n <= 1:
        return n

    prev2 = 0
    prev1 = 1

    for i in range(2, n + 1):
        curr = prev1 + prev2

        prev2 = prev1
        prev1 = curr

    return curr


n = int(input("Enter n: "))

print(nthFibonacci(n))
```
